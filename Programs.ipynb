```python
#Second Largest Element
L1 = [1,2,3,4,55,6,7,8,21]
Max1 = float('-inf')
Max2 = float('-inf')
for num in L1:
    if num > Max1:
        Max2 = Max1
        Max1 = num
    elif num > Max2 and num != Max1:
        Max2 = num
print("The Largest Number", Max1)
print("The Second Largest Number", Max2)
```

```python
#Pallindrome Check
S = input("Enter a String:").lower()
rev = ""
for ch in S:
    rev = ch + rev
if S == rev:
    print("Pallindrome")
else:
    print("Its not a Pallindrome")
```

```python
#Merge array
arr1 = [1,3,5,7,9]
arr2 = [2,4,6,8,10]
i = 0
j = 0
Merged =[]
while i < len(arr1) and j < len(arr2):
    if arr1[i] < arr2[j]:
        Merged.append(arr1[i])
        i += 1
    else:
        Merged.append(arr2[j])
        j += 1
while i < len(arr1):
    Merged.append(arr1[i])
    i += 1
while j < len(arr2):
    Merged.append(arr2[j])
    j += 1
print(Merged)
```
