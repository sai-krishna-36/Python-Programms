```python
arr = [1,2,3,4,5,6,7,8]
target = int(input())
l = 0
r = len(arr)-1
while l < r:
    s = arr[l] + arr[r]
    if s == target:
        print(arr[l],arr[r])
        break
    elif s < target:
        l += 1
    else:
        r -= 1
```
