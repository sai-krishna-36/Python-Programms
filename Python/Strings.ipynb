```python
s = input("Enter a string:")
max_count = 0
max_char = ""
for ch in s:
    count = 0
    for c in s:
        if ch == c:
            count += 1
    if count > max_count:
        max_count = count
        max_char = ch
print(f"The maximum time appered {max_char} : {max_count}")
```

```python
s1 = "Managaraii"
s2 = "garamanam"
li=list(s2)
if len(s1) != len(s2):
    print("Its not a anagram")
else:
    for ch in s1:
        if ch in li:
            li.remove(ch)
        else:
            print("Its not an anagram")
            break
    else:
        print("Yes its an anagram")
```

```python
s = "aaabbcc"
s2 =""
count = 1
for i in range(1,len(s)):
    if s[i] == s[i-1]:
        count += 1
    else:
        s2 += s[i-1] + str(count)
        count = 1
s2 += s[-1] + str(count)
print(s2)
```
