```python
s = input("Enter a string:")
max_count = 0
max_char = ""
for ch in s:
    count = 0
    for c in s:
        if ch == c:
            count += 1
    if count > max_count:
        max_count = count
        max_char = ch
print(f"The maximum time appered {max_char} : {max_count}")
```

```python
s1 = input("Enter a String:")
s2 = input("Enter a String2:")
li = list(s2)
for ch in s1:
    if ch in li:
        li.remove(ch)
    else:
        print("Its not an anagram")
        break
else:
    print("Yes its an anagram")
```
