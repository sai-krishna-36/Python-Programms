```python
arr3 = [42, 17, 12, 89, 3, 56, 73, 24, 91, 65]
right = len(arr3) -1
left = 0
while right > left:
    arr3[left],arr3[right] = arr3[right],arr3[left]
    left += 1
    right -= 1
print(arr3)
```

```python
arr = [0, 1, 0, 3, 12]
pos = 0
for i in range(len(arr)):
    if arr[i] != 0:
        arr[pos], arr[i] = arr[i],arr[pos]
        pos += 1
print(arr)
```

```python
s1 = input("Enter a Sentence: ")
s= ""
count = 0
max_str = ""
max_count = 0
for i in s1:
    if i == " " or i == "," or i == ".":
        if count > max_count :
            max_count = count
            max_str = s
        count = 0
        s = ""
    else:
        count += 1
        s += i
if count > max_count:
    max_count = count 
    max_str = s
print(max_str, max_count)
```

```python
s = input("Enter a String:")
s2 = ""
count = 1
for i in range(1, len(s)):
    if s[i] == s[i-1]:
        count += 1
    else:
        s2 += s[i-1] + str(count)
        count = 1
s2 += s[-1] + str(count)
print(s2)
```
