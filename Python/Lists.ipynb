```python
s = "the sky is blue"

words = []
word = ""

for ch in s:

    if ch != " ":
        word += ch

    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

words.reverse()

result = " ".join(words)

print(result)
```
