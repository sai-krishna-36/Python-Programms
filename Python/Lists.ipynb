```python
s1 = input("Enter a Sentence: ")
s= ""
count = 0
max_str = ""
max_count = 0
for i in s1:
    if i == " " or i == "," or i == ".":
        if count > max_count :
            max_count = count
            max_str = s
        count = 0
        s = ""
    else:
        count += 1
        s += i
if count > max_count:
    max_count = count 
    max_str = s
print(max_str, max_count)
```

```python
s = input("Enter a String:")
s2 = ""
count = 1
for i in range(1, len(s)):
    if s[i] == s[i-1]:
        count += 1
    else:
        s2 += s[i-1] + str(count)
        count = 1
s2 += s[-1] + str(count)
print(s2)
```
