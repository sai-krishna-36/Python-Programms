```python
arr = [4,5,3,4,4,5,3,2,1]
freq = {}
for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)
```

```python
s = "Hello World"
freq2 = {}
for char in s:
    if char.isalpha():
        if char in freq2:
            freq2[char] +=1
        else:
            freq2[char] = 1
print(freq2)
```

```python
s = "aabbcde"

count = {}

for ch in s:
    count[ch] = count.get(ch, 0) + 1
print(count)
for ch in s:
    if count[ch] == 1:
        print(ch,":",count[ch])
```
